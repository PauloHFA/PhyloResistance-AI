# PhyloResistance AI — Explicabilidade (XAI) e Validação Contra o CARD

Este notebook demonstra a camada de **Machine Learning Explicável (TreeSHAP agregado por locus)** em isolados de *Escherichia coli*, respondendo à pergunta científica **Q3**:
> *As explicações do modelo (SHAP) recuperam os determinantes biologicamente conhecidos (CARD/AMRFinderPlus) e sugerem candidatos novos com estabilidade entre dobras?*

```python
import json
from pathlib import Path
import pandas as pd

relatorio_path = Path('../reports/relatorio_validacao.json')
dados = json.loads(relatorio_path.read_text(encoding='utf-8'))
for farmaco, res in dados['resultados_por_farmaco'].items():
    xai = res['explicabilidade_q3']
    print(f'=== {farmaco.upper()} ===')
    print('Proporção Top-6 no CARD:', xai['proporcao_top_k_no_card'])
    print('Determinantes conhecidos recuperados:', xai['determinantes_card_recuperados'])
    print('Candidatos novos (F3 unitigs):', xai['candidatos_novos_top_k'])
    print('Estabilidade Spearman entre dobras:', xai['correlacao_estabilidade_entre_dobras'])
    display(pd.DataFrame(xai['principais_loci_globais']))
```
